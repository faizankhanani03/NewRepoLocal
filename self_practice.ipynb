Ek while loop likhen jo 1 se 10 tak numbers print kare, lekin 5 par ruk jaye. Iske liye break statement ka istemal karen.

```python
i = 0
while i < 10:
    i += 1
    if i == 5:
        break
    print(i ,end=" ")
```
