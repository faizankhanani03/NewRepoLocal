### 1- Find record if Sales greater than 500 and Quantity greater than 5?

```python
import pandas as pd

df = pd.read_csv("retail supermarket.csv")

df1 = df[(df['Sales'] > 500 )  & (df['Quantity'] > 5 )]  # why itny brackets or tuples
df1.head(5)
```
