```python
import pandas as pd 
import numpy as np
import warnings
warnings.filterwarnings('ignore')
df = pd.read_csv('Salaries.csv')
df.drop(['Id'], axis=1, inplace=True)
df.tail(15)
```

Drop rows where EmployeeName and JobTitle are 'Not provided'

```python
df.shape
```

```python
df = df[(df['EmployeeName'].str.lower() != 'Not provided') & (df['JobTitle'].str.lower() != 'Not provided')]

df.shape
```

```python
df.iloc[100:115]
```

Replace 'Not Provided' with 0 in BasePay OvertimePay	OtherPay	Benefits #why not get mean rather then get 0

```python
df['BasePay'] = df['BasePay'].replace('Not Provided', 0)
df['OvertimePay'] = df['OvertimePay'].replace('Not Provided', 0)
df['OtherPay'] = df['OtherPay'].replace('Not Provided', 0)
df['Benefits'] = df['Benefits'].replace('Not Provided', 0)
df.iloc[1000:5000].head()
```

Change BasePay	OvertimePay	OtherPay	Benefits type to float

```python
df['BasePay'] = df['BasePay'].astype(float)
df['OtherPay'] = df['OtherPay'].astype(float)
df['OvertimePay'] = df['OvertimePay'].astype(float)
df['Benefits'] = df['Benefits'].astype(float)

df['OvertimePay'].dtype
```

Display all employees name and basepay columns who have a BasePay greater than 200,000.

```python
df1 = df[df['BasePay'] > 200000][['EmployeeName' , 'BasePay']]
df1.shape
```

```python
df4 = (df[df['TotalPay'] > 300000 ].groupby('JobTitle')['TotalPay'].sum())
df4.head((20))
```

Find the top 5 JobTitles with the highest average TotalPayBenefits show in descending order

```python
grouped = df.groupby('JobTitle')['TotalPayBenefits']
avg = grouped.mean()
s_avg = avg.sort_values(ascending=False)
s_avg.head(5)
```
