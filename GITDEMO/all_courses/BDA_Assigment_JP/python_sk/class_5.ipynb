```python
#  mini project

marks = []

for i in range(5):
    n = int(input("enter stu marks: "))
    marks.append(n)

total = sum(marks)
print(total)

average = total / len(marks)
print(average)

highest = max(marks)
print(highest)
lowest = min(marks)
print(lowest) # i write code why bcz mujhe code to aye

print("\n===== Grades =====")
for i, m in enumerate(marks, 1):
    if m >= 80:
        grade = "A"
    elif m >= 70:
        grade = "B"
    elif m >= 60:
        grade = "C"
    else:
        grade = "F"
    print(f"Student {i}: {m} → Grade: {grade}")
    

```
