```python
# 8️⃣ Student Grade Checker (Dict + Conditionals)

marks = {
    "Ali" : 81,  #, kyu need hai 
    "Ahmed" : 71,            #ye pura if loop me kry to kiya effeiency askati hai code me
    "Bilal" : 61,
    "anas" : 51,
    "abc" : 41,
    "xyz" : 31
}

uer = input("enter stydent to find a grade:").capitalize()

if uer in marks:
    score = marks[uer]
    if score >= 80:
        print(uer,"Grade: A")
    elif score >= 60:
        print(uer , "Grade: B")
    elif score >= 40:
        print(uer, "Grade: C")
    else:
        print(uer,"fail")

else:
    print("student not found:" , uer)

```

```python
# 9️⃣ Vowel Counter Using Dictionary



vowel = {
    "a": 0,
    "e": 0, 
    "i": 0, 
    "o": 0, 
    "u": 0
}

wrd = input("enter your name")

vowel["a"] = wrd.count("a")
vowel["e"] = wrd.count("e")
vowel["i"] = wrd.count("i")
vowel["o"] = wrd.count("o")
vowel["u"] = wrd.count("u")

print(vowel)
print(type(vowel))
```

## Mini Project

```python
store = {
    "apple": 100,
    "banana": 80,
    "milk": 200,
    "bread": 150
}

discounts = {"apple", "milk"}

item = input("Enter product name: ").lower()


if item in store:
    orignal_price = store[item]
    print('price', orignal_price)

    if item in discounts:
        final_price = orignal_price - (orignal_price * 0.10)
        print('final price after 10% discount' , final_price)
    else:
        print("No discount available for this product.")


else:
    price = int(input("Enter price of new product:"))
    store[item] = price

    dis = input("Should this product have a discount?").lower()
    if dis == "yes":
        discounts.add(item)
        print("updated discount set:",discounts)

print("\nUpdated store:", store)
print("Updated discounts:", discounts)
```
