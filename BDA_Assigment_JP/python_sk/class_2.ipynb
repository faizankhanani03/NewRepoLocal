```python
# User se naam lo, aur check karo:

# Agar naam 5 letters se chhota hai → print("Short name")

# Agar 5 se 10 letters ke beech hai → print("Medium name")

# Warna → print("Long name") agar uper nh likha ohr bh output ayga

nam = input('enter your name:')

if len(nam) < len(nam):
    print("Short name")
elif len(nam) > len(nam)  and len(nam) <=len (nam):
    print("Medium name")
else:
    print("Long name")
```

```python
# User se koi word lo aur check karo:

# Agar word me vowels (a, e, i, o, u) hain to print karo “This word has vowels”

# Warna print karo “No vowels found”

user = input("enter word to find vowel:")

if user in ("a,e,i,o,u"):
    print('This word has vowels')
else:
    print("no vowels")
```

```python
# Phir user se input lo aur check karo:

# Agar dono sahi hain → “Login successful!”

# Warna → “Invalid username or password!”

username = "faizan"
password = "12345"

u_name = input("enter username: ")
u_pass = input("enter password: ")

if u_name == username and u_pass == password:
    print("Login successful!")
elif u_name != username and u_pass == password:
    print("invalid username")
elif u_name == username and u_pass != password:
    print("invalid password")
else:
    print("Invalid username or password!")
```

```python
# User se age lo aur:

# Agar < 13 → “Child”

# Agar 13–19 → “Teenager”

# Agar 20–60 → “Adult”

# Agar > 60 → “Senior Citizen”

age = int(input("enter your age: "))

if age < 13:
    print('child')
elif age >= 13 and age <=19:
    print('teenager' , age)
elif age >= 20 and age <=60:
    print('adult')
else:
    print('senior citizen')
```
